```python
# Program Konversi Suhu -

nama = "Rivana Hidayati"
nim = "1306625065"

# Header dan Input Identitas
print(f"Program konversi Suhu")
print(f"Nama : {nama}")
print(f"NIM  : {nim}")
print()

# Input Parameter Suhu
suhu_awal = 0
suhu_akhir = 100
selang = 10

print(f"- Suhu awal  = {suhu_awal}")
print(f"- Suhu akhir = {suhu_akhir}")
print(f"- Selang     = {selang}")
print()

# Header Tabel
print("TABEL KONVERSI")
print("+-----+-----------+----------+-----------+")
print(f"| {'No.':<3} | {'Celcius': <9} | {'Reamur': <8} | {'Fahrenheit':<8} |")
print("+-----+-----------+----------+-----------+")

# Perulangan untuk Menghitung dan Menampilkan Data
celcius = suhu_awal
no = 1

while celcius <= suhu_akhir:
  # Rumus Konversi Suhu
  reamur     = celcius * (4/5)
  fahrenheit = (celcius * (9/5)) + 32

  # Menampilkan baris tabel yang dipisah dengan garis vertikal
  print(f" {no:<4} | {celcius:<9} | {reamur: <8.2f} | {fahrenheit:(<10.2f}|")

  #Update nilai untuk iterasi berikutnya
  celcius += selang
  no += 1

# Garis Penutup Tabel
print("+-----+-----------+----------+-----------+")
```
